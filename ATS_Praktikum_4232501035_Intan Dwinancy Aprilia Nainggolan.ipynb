```python
# ------------------------------------------------------------
# SOAL 6. Status Operasi Boiler
# ------------------------------------------------------------

# Membaca suhu dan tekanan dari input
suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan uap (bar): "))

# Menentukan status operasi boiler
if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")
```

```python
# ------------------------------------------------------------
# SOAL 8. Pengosongan Tangki Bahan Bakar
# ------------------------------------------------------------

volume = 100

# Mengosongkan tangki secara bertahap
while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")

    # Memberikan peringatan saat volume mencapai 20 liter
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Pesan setelah tangki kosong
print("TANGKI KOSONG - Pengosongan berhenti.")
```

SOAL 8. Pengosongan Tangki Bahan Bakar

Cara Kerja Program:

Pada program ini, volume awal tangki ditentukan sebesar 100 liter. Proses pengosongan dilakukan menggunakan perulangan while selama volume masih lebih dari 0. Setiap perulangan akan mengurangi volume sebanyak 20 liter, kemudian menampilkan volume yang tersisa. Program juga memeriksa apakah volume sudah mencapai 20 liter. Jika iya, pesan peringatan akan ditampilkan untuk menunjukkan bahwa tangki hampir kosong. Ketika volume sudah mencapai 0 liter, kondisi perulangan tidak lagi terpenuhi sehingga proses berhenti dan program menampilkan pesan bahwa tangki telah kosong.


Fungsi Kode:
1. volume = 100, menentukan volume awal tangki sebesar 100 liter.
2. while volume > 0, menjalankan perulangan selama volume masih lebih dari 0.
3. volume -= 20, mengurangi volume tangki sebanyak 20 liter setiap perulangan.
4. print(), menampilkan volume tangki dan pesan peringatan.
5. if volume == 20, memeriksa apakah volume tangki sudah mencapai tepat 20 liter.
6. ==, membandingkan dua nilai untuk mengetahui apakah keduanya sama.
7. print("TANGKI KOSONG - Pengosongan berhenti."), menampilkan pesan setelah proses pengosongan selesai.

```python
# ------------------------------------------------------------
# SOAL 10. Pembacaan Suhu Bearing dengan Sensor Rusak
# ------------------------------------------------------------

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

# Memeriksa setiap data suhu bearing
for nomor, nilai in enumerate(data, start=1):
    try:
        suhu_bearing = float(nilai)
        jumlah_valid += 1

        # Memeriksa apakah suhu bearing melebihi 70 derajat
        if suhu_bearing > 70.0:
            print(f"Warning: Titik {nomor} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {nomor}: data tidak valid - pemeriksaan dihentikan")
        break

# Menampilkan jumlah data valid
print(f"Jumlah data valid: {jumlah_valid}")
```

SOAL 10. Pembacaan Suhu Bearing dengan Sensor Rusak

Cara Kerja Program:

Pada program ini, data suhu bearing disimpan dalam list dengan tipe data teks. Pemeriksaan dilakukan menggunakan perulangan for untuk membaca setiap data satu per satu. Fungsi enumerate() digunakan agar setiap data memiliki nomor titik pengukuran yang dimulai dari 1. Selanjutnya, setiap data diubah menjadi float menggunakan try-except. Jika data berhasil diubah, jumlah data valid akan bertambah dan program memeriksa apakah suhu melebihi 70°C. Jika suhu melebihi batas tersebut, program menampilkan peringatan. Namun, ketika ditemukan data "N/A", proses konversi akan menghasilkan ValueError. Program kemudian menampilkan pesan bahwa data tidak valid dan menggunakan break untuk menghentikan perulangan. Setelah itu, jumlah data valid yang berhasil diperiksa ditampilkan.

Fungsi Kode:
1. data = ["62.5", "58.0", "81.3", "N/A", "66.0"], menyimpan data suhu bearing dalam bentuk list.
2. jumlah_valid = 0, menentukan nilai awal jumlah data valid.
3. for, mengulang pemeriksaan setiap data dalam list.
4. enumerate(data, start=1), mengambil data sekaligus memberikan nomor urut yang dimulai dari 1.
5. try, mencoba menjalankan proses konversi data.
6. float(nilai), mengubah data teks menjadi bilangan desimal.
7. jumlah_valid += 1, menambah jumlah data valid sebanyak 1.
8. if suhu_bearing > 70.0, memeriksa apakah suhu bearing melebihi 70°C.
9. except ValueError, menangani kesalahan ketika data tidak dapat diubah menjadi angka.
10. break, menghentikan perulangan ketika ditemukan data tidak valid.
11. print(), menampilkan peringatan dan jumlah data valid.

```python
# ------------------------------------------------------------
# CHALLENGE BONUS. Pemantauan Beban Harian Pembangkit
# ------------------------------------------------------------


# Data beban listrik selama lima kali pengukuran
beban_harian = ["320", "450", "520", "610", "480"]

total_beban = 0
jumlah_valid = 0

# Memeriksa setiap data beban listrik
for nomor, nilai in enumerate(beban_harian, start=1):
    try:
        beban = float(nilai)
        total_beban += beban
        jumlah_valid += 1

        # Menentukan status berdasarkan besar beban listrik
        if beban >= 600:
            status = "BEBAN BERLEBIH"
        elif beban >= 500:
            status = "BEBAN TINGGI"
        else:
            status = "BEBAN NORMAL"

        print(f"Pengukuran {nomor}: {beban:g} kW - {status}")

    except ValueError:
        print(f"Pengukuran {nomor}: data tidak valid")

# Menghitung dan menampilkan rata-rata beban
if jumlah_valid > 0:
    rata_rata = total_beban / jumlah_valid
    print(f"Total beban = {total_beban:g} kW")
    print(f"Rata-rata beban = {rata_rata:.2f} kW")
else:
    print("Tidak ada data beban yang valid.")
```

CHALLENGE BONUS. Pemantauan Beban Harian Pembangkit

Cara Kerja Program:

Program ini digunakan untuk memantau beban listrik pembangkit berdasarkan lima data pengukuran. Data disimpan dalam sebuah list, kemudian diperiksa satu per satu menggunakan perulangan for. Setiap data dikonversi menjadi tipe numerik menggunakan float() agar dapat dihitung. Selanjutnya, program menentukan status beban menggunakan percabangan if-elif-else, yaitu beban normal, beban tinggi, atau beban berlebih. Program juga menjumlahkan seluruh beban yang valid dan menghitung nilai rata-ratanya. Jika terdapat data yang tidak dapat dikonversi menjadi angka, try-except akan menangani kesalahan tersebut agar program tetap dapat berjalan.

Fungsi Kode:
1. beban_harian = ["320", "450", "520", "610", "480"] digunakan untuk menyimpan lima data pengukuran beban listrik dalam bentuk list.
2. total_beban = 0 dan jumlah_valid = 0 digunakan untuk menginisialisasi variabel sebelum proses perhitungan dimulai.
3. for nomor, nilai in enumerate(beban_harian, start=1): digunakan untuk memeriksa setiap data beban secara berulang dan memberikan nomor pada setiap pengukuran, dimulai dari angka 1.
4. try: digunakan untuk menjalankan proses yang berpotensi mengalami kesalahan, sedangkan beban = float(nilai) digunakan untuk mengubah data string menjadi angka desimal agar dapat dihitung.
5. total_beban += beban digunakan untuk menjumlahkan seluruh beban listrik, sedangkan jumlah_valid += 1 digunakan untuk menghitung banyaknya data yang berhasil diproses.
6. if beban >= 600:, elif beban >= 500:, dan else: digunakan untuk menentukan status beban listrik berdasarkan nilai pengukuran, yaitu beban berlebih, beban tinggi, atau beban normal.
7. status = "..." digunakan untuk menyimpan hasil pengelompokan status beban sebelum ditampilkan menggunakan print().
8. except ValueError: digunakan untuk menangani kesalahan apabila terdapat data yang tidak dapat dikonversi menjadi angka.
9. if jumlah_valid > 0: digunakan untuk memastikan terdapat data valid sebelum menghitung rata-rata agar tidak terjadi pembagian dengan nol.
10. rata_rata = total_beban / jumlah_valid digunakan untuk menghitung rata-rata beban listrik berdasarkan jumlah data yang valid.
11. print() digunakan untuk menampilkan hasil pengukuran, status beban, total beban, dan rata-rata beban. Format :.2f digunakan agar rata-rata ditampilkan dengan dua angka di belakang koma.
